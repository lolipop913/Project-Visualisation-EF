# Data Collection

Load and document raw datasets: Eurostat English language learning and EF English Proficiency Index.

## Setup

```python
import sys
import pathlib
import pandas as pd
import eurostat
import warnings

warnings.filterwarnings('ignore')

project_root = pathlib.Path().resolve().parent
sys.path.insert(0, str(project_root))

pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
```

## Data Sources

**Eurostat `educ_uoe_lang01`:** % of tertiary students learning English, 2013–2022.  
**EF English Proficiency Index:** Annual proficiency scores, 2012–2022.

## Load Eurostat

```python
eurostat_raw = eurostat.get_data_df('educ_uoe_lang01', flags=False)
print(f'Shape: {eurostat_raw.shape}')
print(f'Columns: {list(eurostat_raw.columns)[:6]}')
eurostat_raw.head()
```

## Load EF Data

```python
ef_path = project_root / 'data' / 'raw' / 'efiepi_rankings.csv'
ef_raw = pd.read_csv(ef_path)
print(f'Shape: {ef_raw.shape}')
print(f'Columns: {list(ef_raw.columns)}')
ef_raw.head(10)
```

## Validation

```python
# Detect Eurostat geographic column dynamically
geo_col = [c for c in eurostat_raw.columns if 'geo' in c.lower()][0]

print('Eurostat:')
print(f'  Years: {sorted([int(c) for c in eurostat_raw.columns if c.isdigit()])[:5]}...')
print(f'  Countries: {eurostat_raw[geo_col].nunique()}')
print(f'  Missing values: {eurostat_raw.isnull().sum().sum()}')

print()

print('EF EPI:')
print(f'  Years: {ef_raw["Year"].min()} to {ef_raw["Year"].max()}')
print(f'  Countries: {ef_raw["Country"].nunique()}')
print(f'  Missing values: {ef_raw.isnull().sum().sum()}')
```
